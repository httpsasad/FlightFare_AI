```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
```

```python
df = pd.read_csv('../data/Clean_Dataset.csv')
print(df.shape)
df.head()
```

```python
df.info()
df.describe(include='all').T
```

```python
# Remove accidental index columns and duplicates
unnamed = [c for c in df.columns if c.lower().startswith('unnamed')]
df = df.drop(columns=unnamed, errors='ignore').drop_duplicates()

print('Shape:', df.shape)
print('Missing values:\n', df.isna().sum())
```

```python
fig, ax = plt.subplots(figsize=(10,5))
sns.histplot(df['price'], bins=50, kde=True, ax=ax)
ax.set_title('Flight Ticket Price Distribution')
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10,5))
sns.boxplot(data=df, x='class', y='price', ax=ax)
ax.set_title('Price by Class')
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10,5))
sns.boxplot(data=df, x='airline', y='price', ax=ax)
ax.tick_params(axis='x', rotation=45)
ax.set_title('Price by Airline')
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10,5))
sns.scatterplot(data=df.sample(min(10000, len(df)), random_state=42), x='days_left', y='price', alpha=.25, ax=ax)
ax.set_title('Price vs Days Left')
plt.show()
```
